```python
print("Welcome to the Magic 8 ball!") # welcome to game
print("Type 'quit' at any time to exit")


name = input("Enter your name: ") # ask user to enter name
print(f"Hello, {name}!")


import random # 8-ball responces are random from list -->

answer = [
    "without a doubt.",
    "outlook good.",
    "signs point to yes",
    "it is certain.",
    "Definitely.",
    "Concentrate and ask again.",
    "My sources say no.",
    "Don't count on it.",
    "Very doubtful.",
    "absolutly not.",
    "No.",
]


question = input("Enter your question: ") # ask user to enter yes or no question
print("Reading your future...") # loading


if question == 'quit': # if user wants to quit
  print("Your destiny awaits. Farewell!")

elif not question: # if user does not enter a question
  print("You must ask a question to recieve an answer.")

else:
  print(random.choice(answer))
```
